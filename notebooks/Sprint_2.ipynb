# SPRINT 2: Model Building and Evaluation

## Objective

The goal of Sprint 2 is to build, train, and evaluate multiple baseline regression models using the preprocessed datasets generated in Sprint 1.

This sprint focuses on comparing different regression algorithms, evaluating their performance using multiple metrics, identifying overfitting or underfitting, and selecting the best-performing baseline models for optimization in Sprint 3.

This sprint includes:

- Loading the preprocessed datasets
- Building a baseline model
- Training multiple regression models
- Evaluating model performance
- Checking for overfitting and underfitting
- Comparing model performance
- Selecting the best baseline models

```python
import warnings
warnings.filterwarnings("ignore")

# Data Manipulation
import pandas as pd
import numpy as np

# Machine Learning Models
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR

# Evaluation Metrics
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

##  Step 1: Load Preprocessed Data

```python
#Loading Data
X_train = pd.read_csv("../dataset/processed/X_train_processed.csv")
X_test = pd.read_csv("../dataset/processed/X_test_processed.csv")

y_train = pd.read_csv("../dataset/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../dataset/processed/y_test.csv").squeeze()
```

```python
#Verify Dataset Shapes
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("\nTraining Target :", y_train.shape)
print("Testing Target  :", y_test.shape)
```

```python
#Display Dataset Information
print("Training Features")
display(X_train.head())

print("Testing Features")
display(X_test.head())
```

```python
#Verify Missing Values
print("Missing Values in X_train")
print(X_train.isnull().sum().sum())

print("\nMissing Values in X_test")
print(X_test.isnull().sum().sum())

print("\nMissing Values in y_train")
print(y_train.isnull().sum())

print("\nMissing Values in y_test")
print(y_test.isnull().sum())
```

Observations

- Successfully loaded the preprocessed training and testing datasets generated in Sprint 1.
- Verified the dimensions of the feature and target datasets.
- Confirmed that there are no missing values in the datasets.
- The datasets are ready for training and evaluating multiple regression models.

## Step 2: Baseline Model

The objective of this step is to build a baseline regression model using **Linear Regression**. A baseline model serves as a benchmark against which more advanced machine learning models can be compared.

Linear Regression is simple, fast, and highly interpretable, making it an ideal starting point for regression problems.

```python
#Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
lr_train_pred = lr.predict(X_train)
lr_test_pred = lr.predict(X_test)
```

```python
lr_train_mae = mean_absolute_error(y_train, lr_train_pred)
lr_test_mae = mean_absolute_error(y_test, lr_test_pred)

lr_train_rmse = np.sqrt(mean_squared_error(y_train, lr_train_pred))
lr_test_rmse = np.sqrt(mean_squared_error(y_test, lr_test_pred))

lr_train_r2 = r2_score(y_train, lr_train_pred)
lr_test_r2 = r2_score(y_test, lr_test_pred)
```

```python
linear_regression_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R² Score"],
    "Training": [
        lr_train_mae,
        lr_train_rmse,
        lr_train_r2
    ],
    "Testing": [
        lr_test_mae,
        lr_test_rmse,
        lr_test_r2
    ]
})

linear_regression_results = linear_regression_results.round(4)

display(linear_regression_results)
```

```python
dt_train_mae = mean_absolute_error(y_train, dt_train_pred)
dt_test_mae = mean_absolute_error(y_test, dt_test_pred)

dt_train_rmse = np.sqrt(mean_squared_error(y_train, dt_train_pred))
dt_test_rmse = np.sqrt(mean_squared_error(y_test, dt_test_pred))

dt_train_r2 = r2_score(y_train, dt_train_pred)
dt_test_r2 = r2_score(y_test, dt_test_pred)
```

```python
#SVR
svr = SVR()

svr.fit(X_train, y_train)

svr_train_pred = svr.predict(X_train)
svr_test_pred = svr.predict(X_test)
```

```python
svr_train_mae = mean_absolute_error(y_train, svr_train_pred)
svr_test_mae = mean_absolute_error(y_test, svr_test_pred)

svr_train_rmse = np.sqrt(mean_squared_error(y_train, svr_train_pred))
svr_test_rmse = np.sqrt(mean_squared_error(y_test, svr_test_pred))

svr_train_r2 = r2_score(y_train, svr_train_pred)
svr_test_r2 = r2_score(y_test, svr_test_pred)
```

```python
# Random Forest Regressor
rf = RandomForestRegressor(random_state=42)

rf.fit(X_train, y_train)

rf_train_pred = rf.predict(X_train)
rf_test_pred = rf.predict(X_test)
```

```python
rf_train_mae = mean_absolute_error(y_train, rf_train_pred)
rf_test_mae = mean_absolute_error(y_test, rf_test_pred)

rf_train_rmse = np.sqrt(mean_squared_error(y_train, rf_train_pred))
rf_test_rmse = np.sqrt(mean_squared_error(y_test, rf_test_pred))

rf_train_r2 = r2_score(y_train, rf_train_pred)
rf_test_r2 = r2_score(y_test, rf_test_pred)
```

```python
# Gradient Boosting Regressor
gbr = GradientBoostingRegressor(random_state=42)

gbr.fit(X_train, y_train)

gbr_train_pred = gbr.predict(X_train)
gbr_test_pred = gbr.predict(X_test)
```

```python
gbr_train_mae = mean_absolute_error(y_train, gbr_train_pred)
gbr_test_mae = mean_absolute_error(y_test, gbr_test_pred)

gbr_train_rmse = np.sqrt(mean_squared_error(y_train, gbr_train_pred))
gbr_test_rmse = np.sqrt(mean_squared_error(y_test, gbr_test_pred))

gbr_train_r2 = r2_score(y_train, gbr_train_pred)
gbr_test_r2 = r2_score(y_test, gbr_test_pred)
```

```python
# Ridge Regression
ridge = Ridge()

ridge.fit(X_train, y_train)

ridge_train_pred = ridge.predict(X_train)
ridge_test_pred = ridge.predict(X_test)
```

```python
ridge_train_mae = mean_absolute_error(y_train, ridge_train_pred)
ridge_test_mae = mean_absolute_error(y_test, ridge_test_pred)

ridge_train_rmse = np.sqrt(mean_squared_error(y_train, ridge_train_pred))
ridge_test_rmse = np.sqrt(mean_squared_error(y_test, ridge_test_pred))

ridge_train_r2 = r2_score(y_train, ridge_train_pred)
ridge_test_r2 = r2_score(y_test, ridge_test_pred)
```

```python
# Lasso Regression
lasso = Lasso(random_state=42)

lasso.fit(X_train, y_train)

lasso_train_pred = lasso.predict(X_train)
lasso_test_pred = lasso.predict(X_test)
```

```python
lasso_train_mae = mean_absolute_error(y_train, lasso_train_pred)
lasso_test_mae = mean_absolute_error(y_test, lasso_test_pred)

lasso_train_rmse = np.sqrt(mean_squared_error(y_train, lasso_train_pred))
lasso_test_rmse = np.sqrt(mean_squared_error(y_test, lasso_test_pred))

lasso_train_r2 = r2_score(y_train, lasso_train_pred)
lasso_test_r2 = r2_score(y_test, lasso_test_pred)
```

## Step 4: Model Evaluation

The objective of this step is to evaluate the performance of each regression model using multiple evaluation metrics. These metrics help assess prediction accuracy, compare model performance, and identify the most suitable baseline model for further optimization.

```python
model_comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest",
        "Gradient Boosting",
        "Ridge Regression",
        "Lasso Regression",
        "Support Vector Regression"
    ],

    "Train MAE": [
        lr_train_mae,
        dt_train_mae,
        rf_train_mae,
        gbr_train_mae,
        ridge_train_mae,
        lasso_train_mae,
        svr_train_mae
    ],

    "Test MAE": [
        lr_test_mae,
        dt_test_mae,
        rf_test_mae,
        gbr_test_mae,
        ridge_test_mae,
        lasso_test_mae,
        svr_test_mae
    ],

    "Train RMSE": [
        lr_train_rmse,
        dt_train_rmse,
        rf_train_rmse,
        gbr_train_rmse,
        ridge_train_rmse,
        lasso_train_rmse,
        svr_train_rmse
    ],

    "Test RMSE": [
        lr_test_rmse,
        dt_test_rmse,
        rf_test_rmse,
        gbr_test_rmse,
        ridge_test_rmse,
        lasso_test_rmse,
        svr_test_rmse
    ],

    "Train R²": [
        lr_train_r2,
        dt_train_r2,
        rf_train_r2,
        gbr_train_r2,
        ridge_train_r2,
        lasso_train_r2,
        svr_train_r2
    ],

    "Test R²": [
        lr_test_r2,
        dt_test_r2,
        rf_test_r2,
        gbr_test_r2,
        ridge_test_r2,
        lasso_test_r2,
        svr_test_r2
    ]
})

model_comparison = model_comparison.round(4)

display(model_comparison)
```

```python
#Sort by Best Performing Model
model_comparison.sort_values(
    by="Test R²",
    ascending=False
).reset_index(drop=True)
```

```python
#Overfitting Analysis
overfitting_analysis = pd.DataFrame({
    "Model": model_comparison["Model"],
    "Train R²": model_comparison["Train R²"],
    "Test R²": model_comparison["Test R²"]
})

overfitting_analysis["Difference"] = (
    overfitting_analysis["Train R²"] -
    overfitting_analysis["Test R²"]
).round(4)

display(overfitting_analysis)
```

Interpretation Guide

The difference between the **Training R² Score** and **Testing R² Score** indicates how well a model generalizes to unseen data.

| Difference (Train R² − Test R²) | Interpretation |
|:-------------------------------:|----------------|
| < 0.05 | Excellent Generalization |
| 0.05 – 0.10 | Slight Overfitting |
| 0.10 – 0.20 | Moderate Overfitting |
| > 0.20 | Severe Overfitting |

Notes:

- A small difference between the training and testing R² scores indicates that the model generalizes well to unseen data.
- A large positive difference suggests that the model performs significantly better on the training data than on the testing data, indicating overfitting.
- A negative difference (where the testing R² is slightly higher than the training R²) is generally not a concern and may occur due to random train-test splits or normal statistical variation.
- The ideal model achieves a high Testing R² Score while maintaining a minimal difference between training and testing performance.

## Step 5: Model Comparison & Selection

The objective of this step is to compare the performance of all baseline regression models and identify the most suitable candidates for hyperparameter tuning in Sprint 3.

The comparison is based on the evaluation metrics obtained in the previous step, with primary emphasis on **Testing R² Score**, **RMSE**, and the model's ability to generalize to unseen data.

```python
ranking = model_comparison.copy()

ranking["Difference"] = overfitting_analysis["Difference"]

ranking["Test MAE Rank"] = ranking["Test MAE"].rank(method="min")
ranking["Test RMSE Rank"] = ranking["Test RMSE"].rank(method="min")
ranking["Test R² Rank"] = ranking["Test R²"].rank(method="min", ascending=False)
ranking["Generalization Rank"] = ranking["Difference"].abs().rank(method="min")

ranking["Average Rank"] = ranking[
    ["Test MAE Rank", "Test RMSE Rank", "Test R² Rank", "Generalization Rank"]
].mean(axis=1)

ranking = ranking.sort_values(
    by=["Average Rank", "Test R² Rank"]
).reset_index(drop=True)

display(ranking)
```

```python
# Select Top 3 Models for Hyperparameter Tuning
selected_models = ranking.head(3)

display(selected_models)
```

## Model Selection

Based on the evaluation results, the top three regression models were selected for further optimization in Sprint 3.

The selection was based on the following criteria:

- High Testing R² Score
- Low Testing MAE
- Low Testing RMSE
- Strong generalization (minimal difference between Training and Testing R² Scores)

An overall ranking was computed by considering all evaluation metrics instead of relying on a single metric. Based on this comprehensive comparison, the following models demonstrated the best balance between predictive performance and generalization:

- **Random Forest Regressor**
- **Decision Tree Regressor**
- **Gradient Boosting Regressor**

These models will undergo hyperparameter tuning and cross-validation in Sprint 3 to identify the optimal model for deployment.
